# Lab reference

Use this page as a quick decision guide during a lab. Start with the three questions, then open the linked topic notebook for a runnable example.

## 1. I need to

Write the task as a verb: **repeat**, **store**, **search**, **traverse**, **simulate**, **plot**, or **reshape**.

Examples: "repeat until lives run out", "visit every tree node", "simulate many walks", "plot position over time".

## 2. What data do I have?

Check the object before choosing an operation:

- a single value (for example, `lives`)
- a nested object with links (for example, a tree node with `left` and `right`)
- a list or NumPy array; inspect `type(...)`, `.shape`, and a small slice

When in doubt, run `print(data)` or `print(type(data))`.

## 3. Determine the tool

| Need | Data | Tool | Notebook |
| --- | --- | --- | --- |
| Repeat while a condition is true | Number / state | `while` loop | `01_loops_and_state.ipynb` |
| Visit connected nodes | Tree node / `None` | Recursive depth-first search | `02_trees_and_dfs.ipynb` |
| Generate random outcomes | List / NumPy array | `np.random.randint` | `03_numpy_random_walks.ipynb` |
| Estimate chance of finishing at 60+ | Final position from many walks | `np.mean(final_positions >= 60)` | `03_numpy_random_walks.ipynb` |
| Inspect array dimensions | NumPy array | `.shape` | `03_numpy_random_walks.ipynb` |
| Swap rows and columns | 2-D NumPy array | `.T` or `np.transpose(...)` | `03_numpy_random_walks.ipynb` |
| Trigger a rare event with a chosen chance | Probability from 0 to 1 | `np.random.rand() <= probability` | `06_logic_loops_and_simulation.ipynb` |
| Keep fixed related values together | Tuple | `(x, y)` or `(r, g, b)` | `04_lists_arrays_and_indexing.ipynb` |
| Compare many series over time | 2-D NumPy array | `plt.plot(...)` | `03_numpy_random_walks.ipynb` |
| Keep several histories together | List of lists | Nested list, then `np.array(...)` | `04_lists_arrays_and_indexing.ipynb` |
| Look up a value by name | Dictionary | `my_dict[key]` or `.get(key)` | `05_dictionaries_and_pandas.ipynb` |
| Clean, filter, or summarize columns | Pandas DataFrame | `loc`, `iloc`, boolean masks, `groupby` | `05_dictionaries_and_pandas.ipynb` |
| Test conditions or repeat a process | Booleans / sequence | `if`, `while`, `for` | `06_logic_loops_and_simulation.ipynb`

## Fast lab checklist

1. Restate the goal.
2. Print or inspect the data.
3. Choose the smallest tool that matches both.
4. Run a tiny test before scaling up.
5. Label plots and keep the expected output nearby.
