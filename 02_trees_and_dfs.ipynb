# Binary trees and depth-first search

A reference for visiting every node in a binary tree.

## I need to

Visit and print every student in a binary tree, starting at the root.

## What data do I have?

Each `Student` node holds a `name` plus two possible child references: `left` and `right`. A missing child is represented by `None`.

This is linked, hierarchical data—not a flat list—so we follow each child link from a node.

## Determine the tool

Use recursive depth-first search (DFS). The base case returns when the current node is `None`; otherwise, process the node and recurse into its children. This example is **pre-order**: node, left, right.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.left = None
        self.right = None


root = Student("Catalina")
root.left = Student("Emma")
root.right = Student("John")
root.left.left = Student("Alex")


def dfs(node):
    if node is None:  # base case
        return

    print(node.name)
    dfs(node.left)
    dfs(node.right)


dfs(root)
```

## Expected order

`Catalina`, `Emma`, `Alex`, `John`

To change the traversal order, move `print(node.name)` between or after the recursive calls.
