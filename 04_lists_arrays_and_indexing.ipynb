# Lists, NumPy arrays, and indexing

Reference for storing grouped data and selecting the part you need.

## I need to

Keep several related records together, then select one record, one value, or a range of records.

## What data do I have?

`house` is a **list of lists**: each inner list is one room record. Lists can mix types and can be changed. For numerical data with a consistent shape, a 2-D NumPy array is often more convenient.

Think: **each item is itself a list** → list of lists.

## Determine the tool

- Python nested list: `data[row][column]`
- NumPy 2-D array: `data[row, column]`
- One position uses an index; a range uses `start:end`. The start is included and the end is excluded.
- `:` means all entries along that axis.

```python
import numpy as np

house = [["bedroom", 20], ["kitchen", 15], ["bathroom", 8]]

print(house[1])       # one inner list: ['kitchen', 15]
print(house[1][0])    # one value: 'kitchen'
print(house[:2])      # first two rows
print(house[1][0:2]) # a slice of one row

house[2][1] = 9       # lists are mutable
house.append(["hall", 6])
print(house)
```

```python
measurements = np.array([[74, 180], [71, 215], [72, 210]])

print(measurements.shape)  # (rows, columns)
print(measurements[1, 0])  # row 1, column 0
print(measurements[:, 0])  # every row, first column
print(measurements[1, :])  # row 1, every column
print(measurements[-2:, :]) # last two rows
```

## Quick reference

| Goal | List of lists | NumPy array |
| --- | --- | --- |
| One row | `house[1]` | `arr[1, :]` |
| One value | `house[1][0]` | `arr[1, 0]` |
| First two rows | `house[:2]` | `arr[:2, :]` |
| First column | use a loop/comprehension | `arr[:, 0]` |
| Last two rows | `house[-2:]` | `arr[-2:, :]` |

Use a NumPy array only when the data is suitably numerical. Arrays coerce their contents to one data type, so mixed values such as room names and numbers belong in a Python list or a DataFrame.
