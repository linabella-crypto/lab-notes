# Dictionaries and pandas DataFrames

Reference for looking up named values and working with spreadsheet-like data.

## I need to

Look up information by a meaningful name, or clean, select, filter, and summarize table data.

## What data do I have?

- A **dictionary** maps unique keys to values: useful when lookup by name matters.
- A **Series** is one labeled column.
- A **DataFrame** is a 2-D labeled table: rows plus columns.

Choose a Series for one column of information; choose a DataFrame when you need multiple columns together.

## Determine the tool

Use `dictionary[key]` to fetch by key. (`.index()` is a list method, not dictionary lookup.) For pandas, use `[]` for columns, `loc` for label-based selection, `iloc` for position-based selection, and a boolean mask to filter rows.

````python
europe = {
    "spain": {"capital": "madrid", "population": 48.4},
    "france": {"capital": "paris", "population": 68.4},
}

print(europe["spain"]["population"])
europe["norway"] = {"capital": "oslo", "population": 5.5}
print("norway" in europe)
print(europe.keys())
````

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Diana"],
    "Age": [31, 24, 28],
    "Country": ["USA", "UK", "Germany"],
    "Price": [12, 18, 15],
})

display(df)
display(df[["Name", "Age"]])   # multiple columns → DataFrame
print(df["Country"])                # one column → Series
display(df.loc[df["Age"] > 25])     # filter rows
display(df.iloc[1])                    # second row by position
````

````python
# Common clean / analyze operations
df["Country"] = df["Country"].str.upper()
df.loc[df["Name"] == "Bob", "Age"] = 31
df["Name_length"] = df["Name"].apply(len)

print(df.describe())
print(df.groupby("Country")["Price"].sum())
display(df.sort_values("Age", ascending=False))
````

## Useful patterns

```python
df = pd.read_csv("file.csv", index_col=0)
df.loc[:, ["country", "drives_right"]]  # all rows, named columns
df[(df["Age"] > 25) & (df["Age"] < 40)] # both conditions
df[df["Country"].isin(["USA", "Sweden"])]
df.isnull().sum()
df.fillna({"Age": 0})
df.to_csv("cleaned_data.csv", index=False)
```

Use `&`, `|`, and `~` for pandas/NumPy boolean conditions; wrap every condition in parentheses. `and` and `or` are for single Python booleans.
