# Loops and changing state

A reference for repeating work until a condition changes.

## I need to

Repeat an action while there are still lives remaining, then stop and show a final message.

## What data do I have?

- `lives`: one integer that tracks the current state.
- A comparison: `lives > 0`, which becomes either `True` or `False`.

The loop needs a state change; otherwise `lives > 0` will stay true forever.

## Determine the tool

Use a `while` loop when the number of repetitions is not the important part—the stopping condition is. Use `-=` to update the counter each pass.

```python
lives = 3

while lives > 0:
    print(f"You have {lives} lives left. Playing game...")
    lives -= 1

print("Game Over!")
```

## Quick checks

- Start with `lives = 1` to test the last pass.
- If the loop never ends, check that the counter is updated inside the loop.
- If you know the exact number of repetitions in advance, consider a `for` loop instead.
