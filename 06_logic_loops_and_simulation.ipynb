# Logic, loops, and simulation

Reference for making decisions, visiting data, and repeating a random experiment.

## I need to

Decide between outcomes, repeat over a collection, or run the same chance process many times to see its distribution.

## What data do I have?

- One true/false condition → a Python boolean.
- A known collection of items → list, dictionary, array, or DataFrame.
- A changing state with an unknown stopping point → `while` loop.
- Many trial outcomes → collect them in a list, then analyze or plot them.

## Determine the tool

Use `if` / `elif` / `else` for mutually exclusive decisions. Use `for` when stepping through a sequence. Use `while` when a condition controls the stop. Use `np.random` plus a loop when simulating chance.

```python
area = 12
if area > 15:
    print("big place!")
elif area > 10:
    print("medium place!")
else:
    print("pretty small.")

print(area > 10 and area < 18)  # one boolean value
```

```python
areas = [11.25, 18.0, 20.0]
for index, value in enumerate(areas, start=1):
    print(f"Room {index}: {value} sqm")

house = [["bedroom", 20], ["kitchen", 15]]
for room, size in house:
    print(f"The {room} is {size} sqm")

lives = 3
while lives > 0:
    print(f"{lives} lives left")
    lives -= 1
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(123)
final_tails = []

for _ in range(1_000):          # repeat the whole experiment
    tails = [0]
    for _ in range(10):          # one experiment: 10 tosses
        coin = np.random.randint(0, 2)
        tails.append(tails[-1] + coin)
    final_tails.append(tails[-1])

plt.hist(final_tails, bins=10)
plt.xlabel("Number of tails in 10 tosses")
plt.ylabel("Number of simulations")
plt.show()
```

## I need to: give an event a small probability of happening

Use `np.random.rand()` when each trial has a chosen probability of an event. It returns a random decimal from 0 (inclusive) up to 1 (exclusive). Compare it to the probability you want:

- 10% chance → `0.10`
- 1% chance → `0.01`
- 0.5% chance → `0.005`

This is useful for a rare game event, a small chance of a simulation setback, or a random shock in a model—not for selecting a random integer such as a dice roll.

```python
rare_event_probability = 0.10  # 10%

if np.random.rand() <= rare_event_probability:
    print("Rare event happened.")
else:
    print("No rare event this time.")

# Estimate/check the rate over many independent trials:
events = np.random.rand(10_000) <= rare_event_probability
print(f"Observed event rate: {events.mean():.1%}")
```

## Plot choice

- `plt.plot(x, y)`: change over ordered values, often time.
- `plt.scatter(x, y)`: relationship/correlation between two variables.
- `plt.hist(values, bins=...)`: distribution of one variable.

Add `plt.xlabel(...)`, `plt.ylabel(...)`, and `plt.title(...)` before `plt.show()`. Use `plt.clf()` only when you need to clear a figure before creating another one.
